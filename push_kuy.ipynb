# Step by step push:
 1. `git clone <link ssh>` -> clone repo tugas ke lokal
 2. Buat yg sudah ada filenya, move file tsb ke dlm repo lokal. Jika belum ada file nya, silakan buat file assignment di dalam repo lokal.
 3. Edit assignment dan finalisasi, jika sudah yakin atau kalau belum final tapi **mau cicil push** dulu jg gpp silakan `cd <ke repo local>` lalu lanjut ke no 4.
 4. `git status`, untuk cek file apa saja yg berbeda antara repo lokal dengan repo github (warna merah)
 5. `git add .`, . itu artinya semua, kalau mau file tertentu ketikan `git add <nama file>`. Selanjutnya file" tsb akan masuk staging mode
 6. `git status` lagi untuk cek mereka sudah berhasil masuk staging mode (warna hijau)
 7. `git commit -m "<message file>"` untuk memberikan pesan versi, misal pertama push/final push/dll.
 8. `git push origin main`, jika pertama x push, selanjutnya cukup `git push`.
 9. Semisal ada perubahan setelah push, dan mau push ulang, ulangi dari langkah no 4. Tidak perlu git pull dulu karena kamu masih solo (belum team work). Ideal nya jika sudah ada team, biasakan untuk `git pull` dulu baru mulai langkah 4.
 $$end$$
 Tambahan:
 - Jika kita mau meng exclude kan suatu file bisa dengan `.gitignore`, caranya buat file `.gitignore` lalu taruh file yg mau di exclude kan. Nanti file trsbut akan menjadi abu" (artinya sudah tidak akan ikut di push ke repo github)
 - Silakan gunakan `git checkout` untuk cek versi yg mana caranya ketikan `git checkout <nama id version>`. Buat balik ke yg paling update ketikan `git checkout main`.
